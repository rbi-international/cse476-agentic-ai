# Unit 4, Lecture 5: Putting it together, a full system

You built each collaboration primitive alone:
- **L1** fan-out and fan-in (parallel work, gathered with a barrier)
- **L2** routing (send to exactly one, with a default)
- **L3** the manager (delegate by judgement)
- **L4** the blackboard (shared state agents build on)

A real system is **all of them, composed**. This is the finale: one support-triage
system that routes a ticket to a team, fans out parallel checks onto a shared
board, and reads the whole board to decide. Every primitive, one graph, running
**offline**.

## The shape: route, fan out, share, decide

Read the builder and you can see all four primitives: the switch-case is the
router (L2), the fan-out edges are the parallel checks (L1), the fan-in gathers
them (L1), and every node reads and writes the shared board (L4). The decision
step is the manager idea (L3).

```python
from cse476.triage_system import build_triage_system

wf = build_triage_system()
print("full system built:", type(wf).__name__)
print()
print("route (L2) -> fan out (L1) -> share on board (L4) -> fan in + decide (L1+L3)")
```

## A ticket through the whole system

```python
from cse476.triage_system import run_triage_system

# a nasty ticket: billing, a security risk, and urgent
print(await run_triage_system("hacked my account, urgent, I want a refund"))
print()
print("-" * 50)
print()
# a calm ticket
print(await run_triage_system("small question about my invoice"))
```

Read the board line and you have the whole story: routed to a team, security
verdict, priority verdict, and the decision. **Four primitives each wrote one
part of that story, and the final step read all of it.** Because every node is a
plain function, the entire system runs offline and gives the same answer every
time.

## Every primitive is in there

Let's confirm each one contributed to a single run.

```python
out = await run_triage_system("someone hacked it and it is urgent, refund please")
print(out)
print()
print("L2 routing:    ", "team:" in out, "(routed to a team)")
print("L1 fan-out:    ", "security:" in out and "priority:" in out, "(both checks ran)")
print("L4 blackboard: ", out.count(":") >= 4, "(findings accumulated, none lost)")
print("L3 decision:   ", "Decision:" in out, "(final call from the whole board)")
```

## The insight: the blackboard is the connective tissue

```python
from cse476.triage_system import SYSTEM_MAP

for part, role in SYSTEM_MAP.items():
    print(f"{part:26} ->  {role}")
```

Notice what made composition work: the **shared board**. Routing wrote the team
to it, the checks wrote their findings to it, the decision read all of it.
Without shared state, each primitive would be an island. The board is why four
separate primitives behave as **one system** rather than four disconnected steps.
That is why L4 came right before this finale: it is the glue.
