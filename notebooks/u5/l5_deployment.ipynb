## CI/CD: the pipeline is ordered gates

Each stage must pass to continue. The golden rule: a failing test stops the
ship, so a broken change never reaches production.

```python
from cse476.deployment import run_pipeline

# all green -> it ships
print("all pass: ", run_pipeline([("lint", True), ("test", True), ("build", True), ("deploy", True)]))

# a failing test STOPS the ship
print("test fails:", run_pipeline([("lint", True), ("test", False), ("build", True), ("deploy", True)]))
print()
print("-> the broken change never built and never deployed. The gate held.")
print("-> THIS is why the tests from L1 and validation from L2 mattered: they are the gates.")
```
